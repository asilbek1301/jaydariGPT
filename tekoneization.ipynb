```python
text = 'Hello World this is a simple example to show how tokenization works in NLP and a sentence' # 17 token
print(text)
```

```python
tokens = text.split()
print(tokens)
```

```python
vocab = {}
current_id = 0

for token in tokens:
  if token not in vocab:
    vocab[token] = current_id
    current_id += 1

print(vocab)
```

```python
encoded = [vocab[token] for token in tokens]
print(encoded)
```

```python
reverse_vocab = {id: token for token, id in vocab.items()}
decoded = " ".join([reverse_vocab[id] for id in encoded])
print(decoded)
```

```python
# Embedding vector  -> AI Modellar tushunishi uchun yaratilgan        also called as Dense numerical array
# Vector DB         -> Embedding bo'lgan raqamlarni joylashtirish uchun maxsus yaratilgan database
# hello             => [0.41, 0.22, -0.77, 0.56...........]
# hi                => [0.34, 0.72, 0.77, -0.82...........]
```

```python
#-----------------------------------------------------------------------
# PyTorch dan foydalanib Embadding ni amalga oshiramiz
# har bir so'zga o'zi mos keladigan raqamlarni tashkillashtirib beradi
#-----------------------------------------------------------------------

import torch
import torch.nn as nn

embedding_dim = 3
vocab_size = len(vocab)

embedding_layer = nn.Embedding(vocab_size, embedding_dim)

input_ids = torch.tensor([encoded])
embeddings = embedding_layer(input_ids)

print(embeddings)
print(embeddings.shape)
```

```python
from transformers import AutoTokenizer

model_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
print('Vocab size:', tokenizer.vocab_size)
print('Special tokens:', tokenizer.special_tokens_map)
```

```python
text = "Hello how are you doing?"
print('text:', text)
```

```python
tokens = tokenizer.tokenize(text)
print('tokens:', tokens)
```

```python
ids = tokenizer.convert_tokens_to_ids(tokens)
print('token IDs:', ids)
```

```python
encoded = tokenizer(text)
print('encoded:', encoded)
```

```python
print('Input IDs:', encoded['input_ids'])
```

```python
"Hi"                    # => [1, 345, 2, 2, 2] // length 5
"How are you?"          # => [1, 45, 657, 234, 23453] // length 5

# 'pad_token': '</s>'     => tepadagi ikki so'zni ai modelga qo'shib bersa length teng emasligi uchun error qaytaradi,
#                            uning oldini olish uchun pad_token ishlatamiz. fake tokenni (tepadagi 2) qo'shib tenglashtiradi

# 'attention_mask': [1, 1, 1, 1, 1, 1, 1]}    => hamma datalar real
# 'attention_mask': [1, 1, 0, 0, 0]}          => 0 ostidagi datalar fake
```

```python
decoded_text = tokenizer.decode(encoded['input_ids'])
print('decoded_text:', decoded_text)
```

```python
texts = [
    "Hello, how are you?",
    "Today we are learning about tokenizers.",
    "TinyLlama is a small but powerful language model."
]

batch = tokenizer(texts, padding=True, return_tensors="pt")
print(batch.keys())

print("Input IDs shape:", batch["input_ids"].shape)
print('Attention mask shape:', batch["attention_mask"].shape)
```
